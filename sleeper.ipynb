```python
import pandas as pd
from roster import Roster
from tradevalue import TradeValue
```

```python
from tradevalue import TradeValue

TradeValue_Class = TradeValue("https://www.thescore.com/news/2817990",superflex=True)
TradeValue_df = TradeValue_Class.all_dfs
```

```python
TradeValue_df
```

```python
selected_items = ["QB", "RB", "WR", "TE", "base"]
TradeValue_df=TradeValue_df[TradeValue_df["Position"].isin(selected_items)]
player_df=player_df[player_df["position"].isin(selected_items)]
```

```python
TradeValue_df
```

```python
def col_replace(df, column_name):
    df[column_name] = df[column_name].str.replace('.', '', regex=False)
    df[column_name] = df[column_name].str.replace('Kenneth', 'Ken', regex=False)
    df[column_name] = df[column_name].str.replace(' Jr', '', regex=False)
    df[column_name] = df[column_name].str.rstrip(' III')
    df[column_name] = df[column_name].str.rstrip(' II')
    return df
```

```python
TradeValue_df = col_replace(TradeValue_df, 'Player')
player_df = col_replace(player_df, 'full_name')
```

```python
TradeValueIndexDF=TradeValue_df.merge(player_df, left_on='Player', right_on='full_name', how='left')[["Player", "Value", "Position", "index"]]
```

```python
r = requests.get('https://api.sleeper.app/v1/league/1048489707518140416/rosters')
r.json()
```

```python
final_dict = {}
for item in r.json(): 
    for player in item['players']:
        final_dict[player] = item['roster_id']
```

```python
df=pd.DataFrame(final_dict.items(), columns=['player_id', 'roster_id'])
```

```python
TradeValueIndexDF.merge(df, left_on='index', right_on='player_id', how='left').sort_values("Value",ascending=False).to_csv("sleeper.csv", index=False)
```
